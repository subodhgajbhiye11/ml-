```python
import time
from scipy import stats
from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import make_scorer, accuracy_score, f1_score, roc_auc_score, log_loss

# For a CSV with text columns, swap the scaler step for a ColumnTransformer, e.g.:
# prep = make_column_transformer(
#     (OneHotEncoder(handle_unknown="ignore"), cat_cols),
#     remainder=StandardScaler())
# Then wrap it: make_pipeline(prep, LogisticRegression(max_iter=5000))

models = {
    "Logistic Regression": make_pipeline(StandardScaler(),
                                         LogisticRegression(C=1.0, max_iter=5000)),
    "Gradient Boosting": HistGradientBoostingClassifier(max_iter=200,
                                                        learning_rate=0.05,
                                                        random_state=0),
}

scoring = {
    "accuracy": "accuracy",
    "f1": "f1",
    "roc_auc": "roc_auc",
    "log_loss": make_scorer(log_loss, greater_is_better=False,
                            response_method="predict_proba"),
}

cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=42)

results = {}
for name, model in models.items():
    t0 = time.time()
    res = cross_validate(model, X, y, cv=cv, scoring=scoring)
    res["fit_time_total"] = time.time() - t0
    results[name] = res
```

```python
for name, res in results.items():
    print(name)
    print(f"  accuracy: {res['test_accuracy'].mean():.4f} ± {res['test_accuracy'].std():.4f}")
    print(f"  F1:       {res['test_f1'].mean():.4f} ± {res['test_f1'].std():.4f}")
    print(f"  ROC AUC:  {res['test_roc_auc'].mean():.4f} ± {res['test_roc_auc'].std():.4f}")
    print(f"  log loss: {-res['test_log_loss'].mean():.4f} ± {res['test_log_loss'].std():.4f}")
    print(f"  time:     {res['fit_time_total']:.2f}s")

lr, gb = results["Logistic Regression"], results["Gradient Boosting"]
for metric in ["test_accuracy", "test_f1", "test_roc_auc"]:
    t, p = stats.ttest_rel(gb[metric], lr[metric])
    print(f"{metric}: GB - LR = {(gb[metric] - lr[metric]).mean():+.4f}, p = {p:.4f}")

Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=7)
for name, model in models.items():
    model.fit(Xtr, ytr)
    proba = model.predict_proba(Xte)[:, 1]
    pred = (proba >= 0.5).astype(int)
    print(f"{name}: errors = {(pred != yte).sum()}/{len(yte)}, "
          f"log loss = {log_loss(yte, proba):.4f}")
```

```python
import numpy as np, pandas as pd
from scipy import stats
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, log_loss, confusion_matrix, classification_report)

SEED = 42

# 1. Problem and data
data = load_breast_cancer(as_frame=True)
X = data.data
y = (data.target == 0).astype(int)   # positive class = malignant (1)
print("Features:", X.shape[1], "| Samples:", X.shape[0])

# 2. Data inspection
print("Missing values:", int(X.isna().sum().sum()))
print("Duplicate rows:", int(X.duplicated().sum()))
print("Class counts:", y.value_counts().sort_index().to_dict(), "(0=benign, 1=malignant)")

# 3. Stratified 80/20 split, same for both models
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED)
print(f"\nSplit: 80/20 stratified, seed={SEED}")
print(f"Train: {len(X_train)} rows | Test: {len(X_test)} rows")

# 4. Models
models = {
    "Logistic Regression": make_pipeline(StandardScaler(),
                                         LogisticRegression(C=1.0, max_iter=5000)),
    "Gradient Boosting": HistGradientBoostingClassifier(max_iter=200,
                                                        learning_rate=0.05,
                                                        random_state=SEED),
}

# 5. Train on training data only, with CV on training data for stability
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)
for name, m in models.items():
    cv_acc = cross_val_score(m, X_train, y_train, cv=skf, scoring="accuracy")
    print(f"{name} train-CV accuracy: {cv_acc.mean():.4f} ± {cv_acc.std():.4f}")
    m.fit(X_train, y_train)

# 6. Predict the test set once
preds, probs = {}, {}
for name, m in models.items():
    probs[name] = m.predict_proba(X_test)[:, 1]
    preds[name] = (probs[name] >= 0.5).astype(int)

lr, gb = "Logistic Regression", "Gradient Boosting"
sample = pd.DataFrame({
    "actual": y_test.values,
    "LR_pred": preds[lr], "LR_prob": probs[lr].round(3),
    "GB_pred": preds[gb], "GB_prob": probs[gb].round(3),
}, index=y_test.index).head(10)
print("\nFirst 10 test predictions:")
print(sample.to_string())

disagree = np.where(preds[lr] != preds[gb])[0]
print(f"\nModels disagree on {len(disagree)} of {len(y_test)} test rows")
if len(disagree):
    print(pd.DataFrame({"actual": y_test.values[disagree],
                        "LR_prob": probs[lr][disagree].round(3),
                        "GB_prob": probs[gb][disagree].round(3)}).to_string(index=False))

# 7. Metrics and confusion matrices
rows = {}
for name in models:
    p, pr = preds[name], probs[name]
    rows[name] = {
        "Accuracy": accuracy_score(y_test, p),
        "Precision": precision_score(y_test, p),
        "Recall": recall_score(y_test, p),
        "F1": f1_score(y_test, p),
        "ROC AUC": roc_auc_score(y_test, pr),
        "Log loss": log_loss(y_test, pr),
    }
    print(f"\n{name} confusion matrix (rows=actual, cols=predicted):")
    print(pd.DataFrame(confusion_matrix(y_test, p),
                       index=["actual benign", "actual malignant"],
                       columns=["pred benign", "pred malignant"]).to_string())
    print(classification_report(y_test, p, target_names=["benign", "malignant"], digits=4))

# 8. Comparison table
table = pd.DataFrame(rows).T.round(4)
best = {c: (table[c].idxmin() if c == "Log loss" else table[c].idxmax()) for c in table.columns}
print("Comparison table (test set):")
print(table.to_string())
print("\nBest per metric:", best)

# 9. Bootstrap check on the error gap
rng = np.random.default_rng(SEED)
yt = y_test.values
lr_wrong = (preds[lr] != yt); gb_wrong = (preds[gb] != yt)
boot = np.array([gb_wrong[i].mean() - lr_wrong[i].mean()
                 for i in (rng.integers(0, len(yt), len(yt)) for _ in range(5000))])
print(f"\nError-rate gap (GB - LR): {gb_wrong.mean() - lr_wrong.mean():+.4f}")
print(f"Bootstrap 95% CI: [{np.percentile(boot,2.5):+.4f}, {np.percentile(boot,97.5):+.4f}]")
print(f"Share of resamples where LR has fewer errors: {(boot>0).mean():.3f}")
```
